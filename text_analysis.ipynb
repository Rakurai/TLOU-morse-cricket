```python
with open("./data/60min.txt") as f:
    text = f.read()

len(text)
```

```python
def longest_repeated_substring(s):
    n = len(s)
    # Generate all suffixes
    suffixes = [s[i:] for i in range(n)]
    # Sort the suffixes
    suffixes.sort()
    # Find the longest common prefix between consecutive suffixes
    lrs = ""
    for i in range(n - 1):
        j = 0
        while (j < len(suffixes[i]) and j < len(suffixes[i+1]) and 
               suffixes[i][j] == suffixes[i+1][j]):
            j += 1
        if j > len(lrs):
            lrs = suffixes[i][:j]
    return lrs

longest_substring = longest_repeated_substring(text)
print(longest_substring)
```
