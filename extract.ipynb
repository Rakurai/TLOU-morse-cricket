# Fetch YouTube Video and Extract Audio

This notebook downloads a YouTube video and extracts the audio as a WAV file for further analysis.

```python
# Install required packages if needed
# !pip install --upgrade pytube ffmpeg-python
```

```python
from pytube import YouTube
import ffmpeg
import os

# Set the YouTube video URL and output paths
video_url = "https://www.youtube.com/watch?v=EhunEdlmLdg"  # Example: 60min cricket video
output_dir = "./data"
os.makedirs(output_dir, exist_ok=True)
video_path = os.path.join(output_dir, "cricket_video_60min.mkv")
audio_path = os.path.join(output_dir, "cricket_audio.wav")
```

```python
# Download the YouTube video with error handling and troubleshooting tips
try:
    yt = YouTube(video_url)
    print("YouTube object created.")
    stream = yt.streams.filter(only_audio=False, file_extension="mp4").first()
    if stream is None:
        print("No suitable stream found. Available streams:")
        print(yt.streams.filter(file_extension="mp4"))
    else:
        print(f"Downloading stream: {stream}")
        stream.download(output_path=output_dir, filename="cricket_video_60min.mp4")
        print(f"Downloaded video to {video_path}")
except Exception as e:
    print(f"Error downloading video: {e}")
    print("Troubleshooting tips:")
    print("- Ensure the video URL is correct and the video is public.")
    print("- Try updating pytube: pip install --upgrade pytube")
    print("- If the video is age-restricted or private, pytube may not be able to download it.")
    print("- If the error persists, try a different video URL to check if the issue is specific to this video.")
```

```python
# Extract audio and convert to WAV using ffmpeg
ffmpeg.input(video_path).output(audio_path, acodec='pcm_s16le', ac=1, ar='44100').run(overwrite_output=True)
print(f"Extracted audio to {audio_path}")
```

```python
# Split the downloaded video file into separate audio and video streams
audio_only_path = os.path.join(output_dir, "cricket_audio_only.m4a")
video_only_path = os.path.join(output_dir, "cricket_video_only.mp4")

# Extract audio only (no video)
ffmpeg.input(video_path).output(audio_only_path, vn=None, acodec='copy').run(overwrite_output=True)
print(f"Extracted audio-only file to {audio_only_path}")

# Extract video only (no audio)
ffmpeg.input(video_path).output(video_only_path, an=None, vcodec='copy').run(overwrite_output=True)
print(f"Extracted video-only file to {video_only_path}")
```
